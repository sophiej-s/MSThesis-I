```python
#Electronics Dataset:

import numpy as np
import pandas as pd
fileElectr='amazon_reviews_us_Electronics_v1_00.tsv'
df=pd.read_csv(fileElectr, sep="\t", header=0, on_bad_lines='skip')
df=df.dropna(subset=['review_headline', 'review_body', 'star_rating'])
```

```python
#Average Word Length Review Body of a group with the specific star rating
for rating in range(1, 6, 1):

    df_sub=df[df['star_rating']==rating]


    df_sub=df_sub.reset_index() #make it 0-indexed for the interation in the for loop
    word_count=[]

    for index in range(0,df_sub['star_rating'].count() ):
        word_count.append (len (df_sub['review_body'][index].split(" ")) ) #split and count the word

    print("For Rating", rating, ", the mean word count is :", mean(word_count)    )
    
```

```python
#Average Word Length Review Headline of a group with the specific star rating

for rating in range(1, 6, 1):

    df_sub=df[df['star_rating']==rating]


    df_sub=df_sub.reset_index() #make it 0-indexed for the interation in the for loop
    word_count=[]

    for index in range(0,df_sub['review_headline'].count() ):
        word_count.append (len (df_sub['review_headline'][index].split(" ")) ) #split and count the word

    print("For Rating", rating, ", the mean word count is :", mean(word_count)    )
    
```

```python
#put these together to produce a function to be called for other datasets:

def statistics_dataset(df):
    #Total Number of Reviews in the Dataset
    print("\nTotal Number of Reviews in the Dataset")
    print(df['star_rating'].count())
    
    
    #Numbers of Reviews broken down by the Star Value
    print("\nNumbers of Reviews broken down by the Star Value")
    print(df['star_rating'].value_counts())
   

    #Percent of Reviews broken down by the Star Value
    print("\nPercent of Reviews broken down by the Star Value")
    total_num=df['star_rating'].count()
    print(df['star_rating'].value_counts()*100/total_num)

    #Mean Rating
    print("\nMean Rating")
    print(df['star_rating'].mean())


    #Number of Unique Users
    print("\nNumber of Unique Users")
    print(df['customer_id'].nunique())

    #Number of Unique Products
    print("\nNumber of Unique Products")
    print(df['product_id'].nunique())

    #Rating per Product Density (Total Number of Reviews /  Number of Unique Products )
    print("\nRating per Product Density (Total Number of Reviews /  Number of Unique Products )")
    print(df['star_rating'].count()/ (df['product_id'].nunique() ))

    #Rating per User Density (Total Number of Reviews /  Number of Users )
    print("\nRating per User Density (Total Number of Reviews /  Number of Users )")
    print(df['star_rating'].count()/ (df['customer_id'].nunique() ))


    #Number of Unique Products within reviews with 1-5 start ratings
    print("\nNumber of Unique Products within reviews with 1-5 start ratings")
    print(df.groupby('star_rating')['product_id'].nunique())

    
    #Average Word Length Review Body of a group with the specific star rating
    print("\nAverage Word Length Review Body of a group with the specific star rating")
    for rating in range(1, 6, 1):

        df_sub=df[df['star_rating']==rating]


        df_sub=df_sub.reset_index() #make it 0-indexed for the interation in the for loop
        word_count=[]

        for index in range(0,df_sub['star_rating'].count() ):
            word_count.append (len (df_sub['review_body'][index].split(" ")) ) #split and count the word

        print("For Rating", rating, ", the mean word count is :", mean(word_count)    )

        
    #Average Word Length Review Headline of a group with the specific star rating
    print("\nAverage Word Length Headline Body of a group with the specific star rating")
    for rating in range(1, 6, 1):

        df_sub=df[df['star_rating']==rating]


        df_sub=df_sub.reset_index() #make it 0-indexed for the interation in the for loop
        word_count=[]

        for index in range(0,df_sub['review_headline'].count() ):
            word_count.append (len (df_sub['review_headline'][index].split(" ")) ) #split and count the word

        print("For Rating", rating, ", the mean word count is :", mean(word_count)    )

```

```python
#Toys
file_Toys="amazon_reviews_us_Toys_v1_00.tsv"

df_Toys=pd.read_csv(file_Toys, sep="\t", header=0, on_bad_lines='skip')
df_Toys=df_Toys.dropna(subset=['review_headline', 'review_body', 'star_rating'])


statistics_dataset(df_Toys)
```
